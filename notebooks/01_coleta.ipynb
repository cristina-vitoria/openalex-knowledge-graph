Coletamos publicações de instituições brasileiras usando a API do OpenAlex via `pyalex`.

````python
import sys
sys.path.append('..')

from src.fetch import buscar_works, salvar_raw, INSTITUICAO_ROR
````

````python
# Configure sua instituição e o período de interesse
works = buscar_works(
    instituicao_ror=INSTITUICAO_ROR['UFMS'],
    ano_inicio=2018,
    ano_fim=2026,
    field_id=17,
    limite=500,
)

print(f'Total coletado: {len(works)} papers')
````
